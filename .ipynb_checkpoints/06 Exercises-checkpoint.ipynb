## Exercise 06.1 (selecting and passing data structures)

The task in Exercise 04 for computing the area of a triangle involved a function with six arguments ($x$ and $y$ components of each vertex). With six arguments, the likelihood of a user passing arguments in the wrong order is high. 

Use an appropriate data structure, e.g. a `list`, `tuple`, `dict`, etc, to develop a new version of the function with an improved interface (the interface is the arguments that are passed to the function) with fewer arguments. Add appropriate checks inside your function on the input data.

```python
import math
def area(vertex0, vertex1, vertex2):

    #check vertex coordinates are expressed as dictionaries with x and y keys
    if vertex0 != dict(vertex0) and vertex1 != dict(vertex1) and vertex2 != dict(vertex2):
        print("Wrong data type. Vertices must be inputted as dictionaries")
    elif "x" and "y" not in vertex0 and vertex1 and vertex2:
        print("Error. Make sure your vertices include x and y keys to denote coordinates")
        
    #compute area
    value = vertex0["x"]*(vertex1["y"] - vertex2["y"]) + vertex1["x"]*(vertex2["y"] - vertex0["y"]) + vertex2["x"]*(vertex0["y"] - vertex1["y"])
    A = abs(value/2)

    return A

A1 = area({"x": -1.0, "y": 1.0}, {"x": 3.0, "y": 2.0}, {"x": 13.0, "y": 14.0})
assert math.isclose(A1, 19.0)
```

## Exercise 06.2 (selecting data structures)

For a simple (non-intersecting) polygon with $n$ vertices, $(x_0, y_0)$, $(x_1, y_1)$, . . , $(x_{n-1}, y_{n-1})$, the area $A$ is given by
$$
A = \left| \frac{1}{2} \sum_{i=0}^{n-1} \left(x_{i} y _{i+1} - x_{i+1} y_{i} \right) \right|
$$
and where $(x_n, y_n) = (x_0, y_0)$. The vertices should be ordered as you move around the polygon.

Write a function that, given only the vertex coordinates, computes the area of a simple polygon. Test your function for some simple shapes. Pay careful attention to the range of any loops. **Test for polygons without a vertex at the origin.**

```python
#define function. Array containing vertex co-ordinates is expected as parameter
def calc_area(vertices):
    #initialise sum so we can use it later
    sum = 0

    #perform sumatory
    for n  in range(-1, len(vertices) - 1):
        sum += vertices[n][0]*vertices[n + 1][1] - vertices[n + 1][0]*vertices[n][1]

    #calculate area
    area = abs((1/2) * sum)

    return area

#tests
##define shapes and test funtion on each
triangle = [[2, 2], [10, 2], [4, 8]] #area should be 24 square units
print(f"Triangle's area is {calc_area(triangle)}")

pentagon = [[2, 2], [8, 1], [11, 5], [7, 10], [1, 7]] #area should be 58.5 units
print(f"Pentagon's area is {calc_area(pentagon)}")

hexagon = [[2, 3], [6, 1], [11, 3], [12, 8], [7, 11], [1, 8]] #area should be 75.5 units
print(f"Hexagon's area is {calc_area(hexagon)}")
```

## Exercise 06.3 (indexing)

Write a function that uses list indexing to add two vectors of arbitrary length, and returns the new vector. Include a check that the vector sizes match. Print a warning message if there is a size mismatch. The more error information you provide, the easier it would be for someone using your function to debug their code.

Add some tests of your code.

#### Hint: You can create a list of zeros of length `n` by

    z = [0]*n

```python
def sum_vector(x, y):

    #check both vectors are the same length
    if len(x) != len(y):
        diff = len(x) - len(y)
        print("Warning! Vector sizes do not match")
        return "Error"
        #give some extra info to make debugging easier
        if diff < 0:
            print(f"2nd vector has {abs(diff)} more components than the first")
        elif diff > 0:
            print(f"2nd vector has {abs(diff)} less components than the first")

    #Perform vector sum and assign values to a new vector (only if vector lengths match)
    else:
        #use list handling to initialise newvector
        newvector = [0 for i in range(len(x))]
        for i in range(len(x)):
            newvector[i] = x[i] + y[i]

    return newvector

#own tests
x = [1, 15, 2]
y = [8, 9, 0]
z = sum_vector(x, y)
assert z == [9, 24, 2]

d = [1, 2, 3]
e = [12, 0.7]
f = sum_vector(d, e)
assert f == "Error"
```

```python
# tests ##
a = [0, 4.3, -5, 7]
b = [-2, 7, -15, 1]
c = sum_vector(a, b)
assert c == [-2, 11.3, -20, 8]
```

### Optional (advanced) 

Try writing a one-line version that performs the `sum_vector` operation using list comprehension and the built-in function [`zip`](https://docs.python.org/3/library/functions.html#zip).

```python
def better_sum(x, y):
    
    #one-line version using zip. I believe it only works with tuple
    newvector = [a[0] + a[1] for a in list(zip(x, y))]

    return newvector

# repeat test for new algorithm
a = (0, 4.3, -5, 7)
b = (-2, 7, -15, 1)
c = better_sum(a, b)
print(c)
assert c == [-2, 11.3, -20, 8]
```

## Exercise 06.4 (dictionaries)

Create a dictionary that maps college names (the key) to college abbreviations for at least 5 colleges
(you can find abbreviations at https://en.wikipedia.org/wiki/Colleges_of_the_University_of_Cambridge#Colleges).
From the dictionary, produce and print

1. A Python dictionary from college abbreviation to name; and
1. A Python list of college abbreviations sorted into alphabetical order.

*Optional extension:* Create a dictionary that maps college names (the key) to dictionaries of:

- College abbreviation
- Year of foundation 
- Total number students
 
for at least 5 colleges. Take the data from https://en.wikipedia.org/wiki/Colleges_of_the_University_of_Cambridge#Colleges. Using this dictionary, 

1. Find the college with the greatest number of students and print the abbreviation; and 
2. Find the oldest college, and print the number of students and the abbreviation for this college.

```python
## create dictionary of names to abbreviations and initialise others to be filled later
#where the word "college" itself is actually a part of the college name, I have omitted it
c_abbreviations = {"Peterhouse":"PET", "Trinity Hall":"TH", "Selwyn":"SE", "Queens'":"Q", "Darwin":"DAR", "Jesus":"JE", "Clare":"CL"}
abbrev_toname = {}
abbrevs = []

#iterate over both keys and values to fill new dictionary and list
for name, abbrev in c_abbreviations.items():
    abbrev_toname[abbrev] = name
    abbrevs.append(abbrev)

abbrevs.sort()

print(abbrev_toname)
print(abbrevs)
```
